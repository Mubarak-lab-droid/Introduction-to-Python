```python
# --- GLOBAL VARIABLES ---
account_balance = 50000.00
user_pin = "1234"
bank_name = "Zenith Bank"
is_running = True
```

```python
# --- FUNCTIONS ---
```

```python
def display_main_menu():
    """Displays the main USSD options."""
    print(f"\n--- Welcome to {bank_name} USSD Service (*123#) ---")
    print("1. Check Balance")
    print("2. Transfer Money")
    print("3. Buy Airtime")
    print("4. Exit")
```

```python
def check_balance():
    """Handles the balance inquiry feature."""
    entered_pin = input("Enter your 4-digit PIN to view balance: ")
    if entered_pin == user_pin:
        print(f"\n[Success] Your current account balance is: \u20A6{account_balance:,.2f}")
    else:
        print("\n[Error] Ivalid PIN. Transaction canceled.")
        
```

```python
def transfer_money():
    """Handles transferring funds to another account."""
    global account_balance
    print("\n--- Transfer Money ---")
    bank_choice = input("Select Destination Bank:\n1.Apex Bank\n2. Other Banks\nChoice: ")
    account_num = input("Enter Destination Account Number (10 digits): ")

    if len(account_num) != 10 or not account_num.isdigit():
        print("\n[Error] Invalid amount entered.")
        return

    try:
        amount = float(input("Enter amount to transfer; \u20A6"))
    except ValueError:
        print("\n[Error] Invalid amount entered.")
        return

    if amount <= 0:
        print("\n[Error] Amount must be greater than zero.")
    elif amount > account_balance:
        print("\n[Error] Insufficient funds.")
    else:
        entered_pin = input("Enter your 4-digit PIN to confirm: ")
        if entered_pin == user_pin:
            account_balance -= amount
            print(f"\n[Success] \u20A6{amount:,.2f} successfully sent to {account_num}!")
            print(f"Remaining Balance: \u20A6{account_balance:,.2f}")
        else:
            print("\n[Error] Invalid PIN. Transaction canceled.")
```

```python
def buy_airtime():
    """Handles self and third-party airtime top-ups."""
    global account_balance
    print("\n--- Buy Airtime ---")
    option = input("1. Airtime for self\n2. Airtime for Others\nChoice: ")

    if option == "1":
        phone_num = "Registered Line"
    elif option == "2":
        phone_num = input("Enter recipient's phone number: ")
    else:
        print("\n[Error] Invalid choice.")
        return

    try:
        amount = float(input("Enter airtime amount: \u20A6"))
    except ValueError:
        print("\n[Error] Invalid amount.")
        return

    if amount <= 0:
        print("\n[Error] Amount must be greater than zero.")
    elif amount > account_balance:
        print("\n[Error] Insufficient funds.")
    else:
        entered_pin = input("Enter your 4-digit PIN to confirm: ")
        if entered_pin == user_pin:
            account_balance -= amount
            print(f"\n[Success] \u20A6{amount:,.2f} airtime successfully loaded to {phone_num}!")
        else:
            print("\n[Error] Invalid PIN. Transaction canceled.")
```

```python
# --- MAIN LOOP ---
# Simulating dialing the USSD code
dial_code = input("Dial your bank's USSD code (e.g., *123#): ")
```

```python
if dial_code == "*180#":
    while is_running:
        display_main_menu()
        choice = input("\nSelect an option (1-4): ")

        if choice =="1":
            check_balance()
        elif choice =="2":
            transfer_money()
        elif choice =="3":
            buy_airtime()
        elif choice =="4":
            print(f"\nThank you for choosing {bank_name}. Session ended.")
            is_running = False
        else:
            print("\n[Error] Invalid option selected. Please try again.")
else:
    print("Connection problem or invalid MMI code.")
```
