```python
daily_pints_log={1,2,31,2,3,3,4,1,3,5,6,7}

print(daily_pints_log)
```

```python
wifes_daily_pints_log={1,2,3,4,5,3,6,6,24,5,7,0}
```

```python
print(daily_pints_log |wifes_daily_pints_log)
```

```python
print(daily_pints_log & wifes_daily_pints_log)
```

```python
print(daily_pints_log - wifes_daily_pints_log)
```

```python
print(daily_pints_log ^ wifes_daily_pints_log)
```
