```python
import numpy as np
import pandas as pd
```

```python
print("--- Step 1: Generating Sample Data with NumPy ---") 
# Generating dummy transaction IDs and random amounts using NumPy arrays 
transaction_ids = np.arange(1001, 1006) 
amounts = np.array([150.50, 2300.00, 45.10, 890.25, 12000.00]) 
# Generating a random selection of transaction statuses using NumPy 
statuses = np.random.choice(['Success', 'Failed', 'Pending'], size=5)
```

```python
# Combining into a dictionary 
sample_data = { 
"Transaction_ID": transaction_ids, 
"Amount_NGN": amounts, 
"Status": statuses 
} 
print("Data arrays created successfully.")
```

```python
print("\n--- Step 2: Creating and Loading into a Pandas DataFrame ---") 
# Loading the structured sample dictionary into a DataFrame 
df = pd.DataFrame(sample_data) 
```

```python
# Displaying the DataFrame 
print("\nLoaded DataFrame View:") 
print(df)
```

```python
# Displaying basic statistics using Pandas built-in descriptive tools 
print("\n--- Quick Data Summary ---") 
print(f"Total Transaction Volume: ₦{df['Amount_NGN'].sum():,.2f}") 
print(f"Average Transaction Value: ₦{df['Amount_NGN'].mean():,.2f}")
```
